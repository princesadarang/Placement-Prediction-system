```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("placement_dataset.csv")
```

```python
plt.hist(df["cgpa"], bins=10)
plt.xlabel("CGPA")
plt.ylabel("Number of Students")
plt.title("CGPA Distribution")
plt.show()
```

```python
sns.countplot(x="placement_status", data=df)
plt.title("Placement Status")
plt.xlabel("Placement Status")
plt.ylabel("Students")
plt.show()
```

```python
sns.boxplot(x="placement_status", y="cgpa", data=df)
plt.title("CGPA vs Placement")
plt.show()
```

```python
sns.histplot(df["aptitude_score"], kde=True)
plt.title("Aptitude Score Distribution")
plt.xlabel("Aptitude Score")
plt.show()
```

```python
sns.scatterplot(
    x="technical_skill_score",
    y="coding_score",
    data=df
)

plt.title("Technical Skill vs Coding Score")
plt.xlabel("Technical Skill Score")
plt.ylabel("Coding Score")
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.heatmap(
    df.select_dtypes("number").corr(),
    annot=True,
    cmap="coolwarm"
)

plt.title("Feature Correlation Heatmap")
plt.show()
```
